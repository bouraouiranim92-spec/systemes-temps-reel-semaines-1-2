## Partie 1 : Définitions et classification

### 1. Définitions

1. Donnez une définition d'un système temps réel.
2. Expliquez pourquoi un résultat exact mais fourni trop tard peut être considéré comme incorrect.
3. Définissez les termes suivants : période, temps d'exécution, échéance et latence.
4. Expliquez la différence entre une contrainte temporelle stricte et une contrainte temporelle non stricte.

**Réponse :**

1. Un système temps réel est un système informatique qui doit produire des résultats corrects dans un délai déterminé.

2. Un résultat exact fourni trop tard peut être considéré comme incorrect, car il ne répond plus aux contraintes temporelles du système. Par exemple, une commande de freinage envoyée trop tard peut provoquer un accident.

3. Les termes principaux sont :
   - Période (T) : intervalle de temps entre deux activations successives d'une tâche périodique.
   - Temps d'exécution (C) : durée nécessaire au processeur pour exécuter une tâche.
   - Échéance (D) : date limite à laquelle une tâche doit terminer son exécution.
   - Latence : délai entre l'apparition d'un événement et la réponse du système.

4. Une contrainte temporelle stricte doit absolument être respectée : un retard peut entraîner une défaillance grave. Une contrainte temporelle non stricte tolère un certain retard, avec une dégradation possible de la qualité du service.

## Partie 3 : Exercice de calcul

On considère deux tâches périodiques exécutées sur un processeur unique :

- Tâche 1 : `C₁ = 2 ms`, `T₁ = D₁ = 10 ms` ;
- Tâche 2 : `C₂ = 4 ms`, `T₂ = D₂ = 20 ms`.

1. Calculez l'utilisation de chaque tâche.
2. Calculez l'utilisation totale du processeur.
3. Combien d'instances de chaque tâche sont lancées en une seconde ?
4. Les échéances sont-elles respectées si chaque tâche s'exécute dès qu'elle est prête ? Expliquez.
5. Comparez l'utilisation totale à la condition suffisante de Liu et Layland pour deux tâches : `U ≤ 2(2^(1/2) - 1)`.

**Réponse :**


1. Pour une tâche périodique, l'utilisation est :

$$
U_i=\frac{C_i}{T_i}
$$

Pour la tâche 1 :

$$
U_1=\frac{2}{10}=0.2=20\%
$$

Pour la tâche 2 :

$$
U_2=\frac{4}{20}=0.2=20\%
$$

2.


$$
U=U_1+U_2=0.2+0.2=0.4
$$

L'utilisation totale est de **40 %**.

3. Pour la tâche 1 :

$$
N_1=\frac{1000}{10}=100
$$

Pour la tâche 2 :

$$
N_2=\frac{1000}{20}=50
$$

La tâche 1 lance 100 instances par seconde et la tâche 2 lance 50 instances par seconde.

4. Les échéances sont respectées dans cet exemple si le processeur exécute les tâches prêtes sans retard excessif. À l'instant initial, les deux tâches demandent au total :

$$
C_1+C_2=2+4=6\,ms
$$

Leurs échéances sont respectivement à 10 ms et 20 ms. Les deux tâches peuvent donc terminer avant leurs échéances. La vérification complète dépend toutefois de la politique d'ordonnancement utilisée.

5.Pour deux tâches, la condition suffisante est :

$$
U\leq 2(\sqrt{2}-1)
$$

$$
U\leq 0.8284
$$

Or :

$$
0.4\leq 0.8284
$$

La condition est respectée. L'ordonnancement est donc garanti par cette condition suffisante dans le cadre de ses hypothèses, notamment avec l'algorithme Rate Monotonic.


## Partie 4 : Simulation Python

Modifiez les paramètres `C`, `T` et `duree_simulation` dans la cellule suivante. Observez les échéances respectées ou violées.

Testez au minimum les cas suivants :
- `C = 2`, `T = 5` ;
- `C = 5`, `T = 5` ;
- `C = 6`, `T = 5`.

Expliquez dans la cellule Markdown suivante à partir de quelle condition l'ordonnancement devient impossible.

```python
import matplotlib.pyplot as plt

# Paramètres à modifier
C = 2
T = 5
duree_simulation = 30

activations = list(range(0, duree_simulation, T))
fin_execution = [min(activation + C, duree_simulation) for activation in activations]

fig, ax = plt.subplots(figsize=(11, 2.8))
for numero, (activation, fin) in enumerate(zip(activations, fin_execution), 1):
    ax.barh(0.25, fin - activation, left=activation, height=0.2, color="#2f80ed")
    ax.text((activation + fin) / 2, 0.25, f"Instance {numero}", ha="center", va="center", color="white", fontsize=8)
    ax.annotate("", xy=(activation + T, 0.02), xytext=(activation + T, 0.35), arrowprops={"arrowstyle": "<->", "color": "#d64545", "linestyle": "--"})

ax.spines["bottom"].set_position(("data", 0))
ax.spines["top"].set_visible(False)
ax.set_xlim(0, duree_simulation)
ax.set_ylim(-0.1, 0.7)
ax.set_yticks([0.25])
ax.set_yticklabels(["Processeur"])
ax.set_xticks(range(0, duree_simulation + 1, T))
ax.set_xlabel("Temps")
ax.set_title(f"Tâche périodique : C={C}, T={T}")
ax.grid(axis="x", alpha=0.25)
plt.tight_layout()
plt.show()
```

### Interprétation de la simulation

1. Pour quelle relation entre `C` et `T` chaque instance respecte-t-elle son échéance ?
2. Que se passe-t-il lorsque `C > T` ?
3. Pourquoi le cas `C = T` laisse-t-il une marge nulle au processeur ?

**Réponse :**


1. 

Chaque instance respecte son échéance si son temps d'exécution vérifie :

$$
C\leq T
$$

2. 

Lorsque C est supérieur à T, une instance n'a pas terminé avant l'arrivée de l'instance suivante. Les tâches s'accumulent et les échéances ne peuvent plus être respectées durablement sur un processeur unique.

3.

Lorsque C = T, la tâche utilise toute la période disponible. Le processeur ne dispose d'aucune marge pour exécuter d'autres tâches, gérer les interruptions ou absorber un retard.
